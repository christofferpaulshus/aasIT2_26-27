### 1. Logiske operatorer

__Sammenligningsoperatorer.__ Disse operasjonene forteller datamaskinen at den skal ta to verdier (eller variabler) og sammenligne dem. Sammenligningsoperatorene forteller også hvordan disse skal sammenlignes (se tabell under). Resultatet av en sammenligningsoperator er en boolsk verdi.

|Operasjon |Eksempel [kode] |
|:---- |:-----: |
|Er lik |`==` |
|Er ikke lik |`!=` |
|Mindre enn |`<` |
|Større enn |`>` |
|Mindre enn eller lik |`<=` |
|Større enn eller lik |`>=` |

__Logiske operatorer.__ En logisk operator forteller datamaskinen hvordan den skal kombinere flere to ulike sammenligningsoperatorer (se tabell under).

|Operasjon |Eksempel [kode] |
|:---- |:-----: |
|Og (&&) |`and` |
|Eller |`or` |
|Er ikke |`not` |

````python
# Eksempel 1.1 Logiske operatorer

print(10 != 5)
````

### 3. Løkker
En av de store fordelene ved å programmere, er at datamaskinen kan la oss gjenta samme oppgave mange ganger. Mange oppgaver kan være lette å løse, men krever mange gjentakelser. Ved å programmere med løkker kan vi beskrive én gjentakelse, og la datamaskinen ta seg av resten.

Det finnes ulike løkker for forskjellige oppgaver. Vi skal se på de to mest brukte løkkene i Python: for-løkker og while-løkker.

**For-løkker** \
Vi bruker en for-løkke når vi eller programmet vårt vet hvor mange gjentakelser vi ønsker.

<div>
<img src="https://innhold.aunivers.no/image/svg/24408/svg_file/for_font.svg" width="450"/>
</div>

```python
for i in range(10):
    print(i)
```
|```i```|Center|
|:-----:|:----:|
|En variabel. Denne vil forandre verdi for hver gjennomgang av løkka |A |

**While-løkker** \
En while-løkke utfører en bestemt oppgave så lenge en gitt betingelse er sann.

<div>
<img src="https://innhold.aunivers.no/image/svg/24410/svg_file/while_font.svg" width="450"/>
</div>

````python
# Eksempel 3.1 For-løkke

print('hei')
print('hei')
print('hei')
print('hei')
print('hei\n\n')

# enklere å skrive det over som en løkke->

for i in range(1,5+1):
    #print(i)
    print('hei')
````

````python
# Eksempel 3.2 While-løkke

x = 0

while x < 10:
    x = x + 1
    print(x)
````

````python
# Oppgave: print 5 'hei' ved å bruke en while løkke
x = 0

while x < 5:
    x = x + 1
    print('hei')
````

````python
# Eksempel 3.3 Break

x = 0
tall = float(input('Skriv ett tall:'))
max = 1000
p = tall

while True:
    x = x + 1
    p = p * 2

    if p > max:
        print(f'Vi må doble tallet {tall} {x} antall ganger før det blir større enn {max}.')
        break
````

````python
# Eksempel 3.4 Løkke gjennom en streng

s = 'blyant'

for i in s:
    print(i)
````

````python
# Eksempel 3.5 Løkke gjennom en liste

l = ['blyant','viskelær','bok','PC']

for i in l:
    print(i)
````
